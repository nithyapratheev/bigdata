## Step 2 – Start Spark

`SparkSession` is the entry point for working with Spark.

We use `local[*]` so Spark can run locally in the Colab environment.

```python
# Import SparkSession
from pyspark.sql import SparkSession

# Create a Spark session
# local[*] means Spark can use the available local CPU resources
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("BeginnerSalesAnalysis") \
    .getOrCreate()

print("Spark started successfully!")
```

## Step 3 – Create our first RDD

An RDD is a distributed collection of data that Spark can process in parallel.

Here we start with a normal Python list and convert it into an RDD using `parallelize()`.

```python
# A simple Python list containing sales values
sales = [1000, 2000, 1500, 3000, 2500]

# Convert the Python list into an RDD
rdd = spark.sparkContext.parallelize(sales)

# collect() brings the results back so we can see them
print(rdd.collect())
```

## Step 4 – RDD `map()` operation

`map()` applies the same operation to every element.

Below, we increase every sales value by 10%.

```python
# Increase every sales value by 10%
increased_sales = rdd.map(lambda x: x * 1.10)

print(increased_sales.collect())
```

## Step 5 – RDD `filter()` operation

`filter()` keeps only the values that satisfy a condition.

Here we keep sales greater than ₹2,000.

```python
# Keep only sales values greater than 2000
high_sales = rdd.filter(lambda x: x > 2000)

print(high_sales.collect())
```

## Step 6 – RDD `reduce()` operation

`reduce()` combines multiple values to produce one final value.

Here we calculate total sales.

```python
# Add all sales values together
total_sales = rdd.reduce(lambda a, b: a + b)

print("Total Sales:", total_sales)
```

## Step 7 – Create a real-world sales DataFrame

Now we move from a simple RDD example to a structured DataFrame.

The data represents product sales.

```python
# Sample sales data
data = [
    ("Laptop", "Electronics", 50000),
    ("Mobile", "Electronics", 20000),
    ("Laptop", "Electronics", 60000),
    ("Tablet", "Electronics", 30000),
    ("Mobile", "Electronics", 25000),
    ("Laptop", "Electronics", 55000),
    ("Headphones", "Accessories", 5000),
    ("Mouse", "Accessories", 2000)
]

# Column names for the DataFrame
columns = ["Product", "Category", "Sales"]

# Create a PySpark DataFrame
df = spark.createDataFrame(data, columns)

# Display the DataFrame
df.show()
```

## Step 8 – Filter the DataFrame

Find products where the individual sale is greater than ₹30,000.

```python
# Keep rows where Sales is greater than 30000
df.filter(df.Sales > 30000).show()
```

## Step 9 – Find the highest individual sale

`orderBy()` sorts the data. `desc()` sorts from highest to lowest.

```python
# Sort sales from highest to lowest
df.orderBy(df.Sales.desc()).show()
```

## Step 10 – Calculate average sales

We use the `avg()` function to calculate the average sale.

```python
# Import the average function
from pyspark.sql.functions import avg

# Calculate average sales
df.select(avg("Sales").alias("Average_Sales")).show()
```

## Step 11 – Total sales by product

`groupBy()` groups rows with the same product name.

Then `sum()` calculates the total sales for each product.

```python
# Calculate total sales for each product
df.groupBy("Product").sum("Sales").show()
```

## Step 12 – Total sales by category

This is a small Big Data analytics example: we group the sales by category and calculate the total.

```python
# Calculate total sales for each category
df.groupBy("Category").sum("Sales").show()
```
